```python
from transformers import AutoModel, AutoTokenizer

model_name = "HuggingFaceTB/SmolLM2-135M"

tok = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
```

```python
text = "Assalomu alaykum. Bugun xavo juda ham sovuq"
tok.tokenize(text)
```

```python
input_ids = tok(text, return_tensors="pt")["input_ids"]
input_ids
```

```python
embedding = model.embed_tokens
embedding
```

```python
embedding(input_ids)
```

```python
attention_layer = layers[0].self_attn
mlp = layers[0].mlp
input_norm = layers[0].input_layernorm
post_att_norm = layers[0].post_attention_layernorm
```

```python
# Sequential not Parallel
# Attention layer
attention_layer
```

```python
# Har bir token 576 dim vector bo'lib keladi.
# Query(Q) - token nimani qidiryabdi?
# Key (K) - har bir token qanday malumotni taklif qiladi
# Value (V) - token uzatadigan malumot
# Query 576 ta K va Value - 192, Demak har bir queryga 3 tadan K va Value togri kelyabdi. GQA(Grouped query Attention) KV cache MHA ga nisbata 3 barobar kam xotira egallaydi.  
# q_proj: 576 → 576   (Query)
# k_proj: 576 → 192   (Key)
# v_proj: 576 → 192   (Value)
# o_proj: 576 → 576   (Output)
```

```python
# FFN (Feed forward network)
# Har bir token alohida ishlanadi. Attention context beradi, FFN mantiq va faktlarni qollaydi
mlp
```

```python
# Normalization
input_norm
# input_dimensions = 576, har bir token vector hajmi
# eps = 1e-5 prevents division by zero.
```
