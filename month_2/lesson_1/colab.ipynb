```python
import platform
import psutil
import torch

# 1. Print the operating system hosting your code
print(f"🖥️ Cloud OS Platform: {platform.platform()}")

# 2. Check the massive amount of free RAM Google assigned to you
total_ram_gb = psutil.virtual_memory().total / (1024 ** 3)
print(f"💾 Assigned Cloud RAM: {total_ram_gb:.2f} GB")

# 3. Check if your system detects the available GPU resources
gpu_available = torch.cuda.is_available()
print(f"🔥 Cloud GPU Available: {gpu_available} ({torch.cuda.get_device_name(0) if gpu_available else 'None'})")
```
