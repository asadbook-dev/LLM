```python
import json

DATA_FILE = "../data/final/uz_wiki.jsonl"
```

```python
def load_jsonl(path, max_docs=None):
    texts = []
    with open(path, "r", encoding="utf-8") as f:
        for i, line in enumerate(f):
            if max_docs is not None and i >= max_docs:
                break
            texts.append(json.loads(line)["text"])
    return texts

docs = load_jsonl(DATA_FILE)

total_chars = sum(len(t) for t in docs)
total_words = sum(len(t.split()) for t in docs)

print("Hujjatlar    : {:,}".format(len(docs)))
print("Belgilar     : {:,}".format(total_chars))
print("So'zlar      : {:,}".format(total_words))
print("O'rtacha     : {:.0f} belgi / hujjat".format(total_chars / len(docs)))
print()
print("Namuna:")
print(docs[0][:300])
```

```python
from collections import Counter


sample = docs[:20000]                      # o'lchov uchun namuna
sample_chars = sum(len(t) for t in sample)

# --- character-level ---
char_vocab = set()
for t in sample:
    char_vocab.update(t)
# char_tokens = sample_chars                  # har bir belgi = 1 token

# --- word-level ---
word_counter = Counter()
for t in sample:
    word_counter.update(t.split())
word_tokens = sum(word_counter.values())

header = "{:<14}{:>12}{:>16}{:>16}".format("Usul", "Lugat", "Tokenlar", "Token/belgi")
print(header)
print("-" * len(header))
print("{:<14}{:>12,}{:>16,}{:>16.3f}".format(
    "character", len(char_vocab), char_tokens, char_tokens / sample_chars))
print("{:<14}{:>12,}{:>16,}{:>16.3f}".format(
    "word", len(word_counter), word_tokens, word_tokens / sample_chars))
print()

# Word-level lug'atining uzun dumi
singletons = sum(1 for c in word_counter.values() if c == 1)
print("Faqat 1 marta uchragan so'zlar: {:,} / {:,}  ({:.1%})".format(
    singletons, len(word_counter), singletons / len(word_counter)))
print()
print("Eng kam uchraydigan 10 ta so'z:", [w for w, _ in word_counter.most_common()[-10:]])
```

```python
from tokenizers import Tokenizer, models, pre_tokenizers, decoders, trainers


def train_bpe(texts, vocab_size):
    """Train a small byte-level BPE tokenizer on the given texts."""
    tok = Tokenizer(models.BPE())
    tok.pre_tokenizer = pre_tokenizers.ByteLevel(add_prefix_space=False)
    tok.decoder = decoders.ByteLevel()

    trainer = trainers.BpeTrainer(
        vocab_size=vocab_size,
        special_tokens=["<eos>"],
        initial_alphabet=pre_tokenizers.ByteLevel.alphabet(),
        min_frequency=2,
        show_progress=False,
    )
    tok.train_from_iterator(texts, trainer=trainer)
    return tok
```

```python
bpe = train_bpe(sample, vocab_size=8000)

bpe_counter = Counter()
for t in sample:
    bpe_counter.update(bpe.encode(t).tokens)
bpe_tokens = sum(bpe_counter.values())
```

```python
edge_cases = [
    "O'zbekiston",
    "  ikki   probel  ",
    "Birinchi qator\nIkkinchi qator\n\n",
    "Telefon: <PHONE>, email: <EMAIL>",
    "1990-2000 yillar, 3.14, 25%",
    "Ўзбекистон",
    "Salom 😊 dunyo",
    "선배 Sunbae",
    "",
]

for t in edge_cases:
    decoded = bpe.decode(bpe.encode(t).ids)
    status = "OK  " if decoded == t else "FAIL"
    print(f"{status} {t!r:<40} -> {decoded!r}")
```

```python
import random

random.seed(42)
test_docs = random.sample(docs, 1000)

failures = []
for t in test_docs:
    if bpe.decode(bpe.encode(t).ids) != t:
        failures.append(t)

print(f"Passed: {len(test_docs) - len(failures)} / {len(test_docs)}")
assert not failures, f"Round-trip failed for {len(failures)} documents"
print("Round-trip test passed ✅")
```
